# Person 3 - Final Algorithm Portfolio
Completed algorithms: ElasticNet, Decision Tree Classification, Random Forest Regression, SVR, and SVC.
The final file combines both datasets, preprocessing, evaluation, and visualisation.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import ElasticNet
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestRegressor
from sklearn.svm import SVR, SVC
from sklearn.metrics import r2_score, f1_score
housing = pd.read_csv('housing.csv'); X = housing.select_dtypes(include='number').drop(columns='SalePrice'); y = housing.SalePrice
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=.2, random_state=42)
for name, estimator in {'ElasticNet':ElasticNet(alpha=.01, l1_ratio=.5, max_iter=5000), 'Random Forest Regression':RandomForestRegressor(n_estimators=150, random_state=42, n_jobs=-1), 'SVR':SVR(C=100)}.items():
    model = make_pipeline(SimpleImputer(strategy='median'), StandardScaler(), estimator).fit(Xtr, ytr)
    print(name, 'R2=', r2_score(yte, model.predict(Xte)))
shop = pd.read_csv('online_shoppers_intention.csv'); X = shop.select_dtypes(include='number'); y = shop.Revenue.astype(int)
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=.2, stratify=y, random_state=42)
for name, estimator in {'Decision Tree Classification':DecisionTreeClassifier(max_depth=8, random_state=42), 'SVC':SVC(C=2)}.items():
    model = make_pipeline(SimpleImputer(strategy='median'), StandardScaler(), estimator).fit(Xtr, ytr)
    print(name, 'F1=', f1_score(yte, model.predict(Xte)))
housing.plot.scatter(x='Gr Liv Area', y='SalePrice', title='Final regression visualisation'); plt.show()
```
