# Person 3 - Tree and Random Forest Regression
Compare a single decision tree with an ensemble regressor.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score
df = pd.read_csv('housing.csv'); X = df.select_dtypes(include='number').drop(columns='SalePrice'); y = df.SalePrice
X = SimpleImputer(strategy='median').fit_transform(X); Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=.2, random_state=42)
for name, model in {'Decision Tree':DecisionTreeRegressor(max_depth=10, random_state=42), 'Random Forest':RandomForestRegressor(n_estimators=150, random_state=42, n_jobs=-1)}.items():
    model.fit(Xtr, ytr); print(name, r2_score(yte, model.predict(Xte)))
```
