# Person 3 - Preprocessing and Visualisation
Prepare scaled numeric matrices and inspect feature relationships for both tracks.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
housing = pd.read_csv('housing.csv'); shoppers = pd.read_csv('online_shoppers_intention.csv')
housing_X = StandardScaler().fit_transform(SimpleImputer(strategy='median').fit_transform(housing.select_dtypes(include='number').drop(columns='SalePrice')))
shopper_X = StandardScaler().fit_transform(SimpleImputer(strategy='median').fit_transform(shoppers.select_dtypes(include='number')))
sns.histplot(data=shoppers, x='PageValues', hue='Revenue', element='step'); plt.title('Page values and purchase intent'); plt.show()
sns.scatterplot(data=housing, x='Overall Qual', y='SalePrice'); plt.show()
```
