# Person 3 - SVR and SVC
Use support vector methods for housing regression and purchase classification.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVR, SVC
from sklearn.metrics import r2_score, f1_score
housing = pd.read_csv('housing.csv'); X = housing.select_dtypes(include='number').drop(columns='SalePrice'); y = housing.SalePrice
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=.2, random_state=42)
svr = make_pipeline(SimpleImputer(strategy='median'), StandardScaler(), SVR(C=100, epsilon=.1)).fit(Xtr, ytr)
print('SVR R2=', r2_score(yte, svr.predict(Xte)))
shop = pd.read_csv('online_shoppers_intention.csv'); X = shop.select_dtypes(include='number'); y = shop.Revenue.astype(int)
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=.2, stratify=y, random_state=42)
svc = make_pipeline(SimpleImputer(strategy='median'), StandardScaler(), SVC(C=2)).fit(Xtr, ytr)
print('SVC F1=', f1_score(yte, svc.predict(Xte)))
```
