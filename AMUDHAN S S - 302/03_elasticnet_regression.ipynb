# Person 3 - ElasticNet Regression
Combine L1 and L2 regularisation for the housing price target.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import ElasticNet
from sklearn.metrics import r2_score, mean_absolute_error
df = pd.read_csv('housing.csv'); X = df.select_dtypes(include='number').drop(columns='SalePrice'); y = df.SalePrice
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=.2, random_state=42)
model = make_pipeline(SimpleImputer(strategy='median'), StandardScaler(), ElasticNet(alpha=.01, l1_ratio=.5, max_iter=5000)).fit(Xtr, ytr)
pred = model.predict(Xte); print('R2=', r2_score(yte, pred), 'MAE=', mean_absolute_error(yte, pred))
```
