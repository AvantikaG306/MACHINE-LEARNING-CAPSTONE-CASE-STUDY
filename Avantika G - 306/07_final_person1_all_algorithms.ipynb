# Person 1 - Final Algorithm Portfolio
All assigned algorithms are completed: Linear Regression, Ridge, Logistic Regression, KNN Regression, and KNN Classification.
This final notebook consolidates preprocessing, visualisation, training, and evaluation.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, Ridge, LogisticRegression
from sklearn.neighbors import KNeighborsRegressor, KNeighborsClassifier
from sklearn.metrics import r2_score, f1_score
housing = pd.read_csv('housing.csv'); shoppers = pd.read_csv('online_shoppers_intention.csv')
housing_X = housing.select_dtypes(include='number').drop(columns='SalePrice'); housing_y = housing.SalePrice
Xtr, Xte, ytr, yte = train_test_split(housing_X, housing_y, test_size=.2, random_state=42)
regressors = {'Linear Regression':LinearRegression(), 'Ridge':Ridge(alpha=10), 'KNN Regression':KNeighborsRegressor(n_neighbors=7)}
for name, estimator in regressors.items():
    model = make_pipeline(SimpleImputer(strategy='median'), StandardScaler(), estimator).fit(Xtr, ytr)
    print(name, 'R2=', r2_score(yte, model.predict(Xte)))
shop_X = shoppers.select_dtypes(include='number'); shop_y = shoppers.Revenue.astype(int)
Xtr, Xte, ytr, yte = train_test_split(shop_X, shop_y, test_size=.2, stratify=shop_y, random_state=42)
for name, estimator in {'Logistic Regression':LogisticRegression(max_iter=1000), 'KNN Classification':KNeighborsClassifier(n_neighbors=7)}.items():
    model = make_pipeline(SimpleImputer(strategy='median'), StandardScaler(), estimator).fit(Xtr, ytr)
    print(name, 'F1=', f1_score(yte, model.predict(Xte)))
housing.plot.scatter(x='Gr Liv Area', y='SalePrice', title='Final regression visualisation'); plt.show()
```
