# Person 1 - Logistic Regression and KNN Classification
Classification milestone using the online shopping purchase target.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, f1_score
df = pd.read_csv('online_shoppers_intention.csv')
X = df.drop(columns='Revenue'); y = df['Revenue'].astype(int)
cats = X.select_dtypes(exclude='number').columns; nums = X.select_dtypes(include='number').columns
prep = ColumnTransformer([('num', SimpleImputer(strategy='median'), nums), ('cat', OneHotEncoder(handle_unknown='ignore'), cats)])
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=.2, stratify=y, random_state=42)
for name, estimator in {'Logistic': LogisticRegression(max_iter=1000), 'KNN': KNeighborsClassifier(n_neighbors=7)}.items():
    pipe = Pipeline([('prep', prep), ('scale', StandardScaler(with_mean=False)), ('model', estimator)])
    pipe.fit(X_train, y_train); pred = pipe.predict(X_test)
    print(name, accuracy_score(y_test, pred), f1_score(y_test, pred))
```
