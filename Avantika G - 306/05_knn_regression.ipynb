# Person 1 - KNN Regression
Apply distance-based regression after scaling numeric housing features.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import r2_score, mean_squared_error
df = pd.read_csv('housing.csv'); X = df.select_dtypes(include='number').drop(columns='SalePrice'); y = df.SalePrice
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=.2, random_state=42)
for k in [3, 5, 9]:
    model = make_pipeline(SimpleImputer(strategy='median'), StandardScaler(), KNeighborsRegressor(n_neighbors=k))
    model.fit(X_train, y_train); pred = model.predict(X_test)
    print('k=', k, 'R2=', r2_score(y_test, pred), 'RMSE=', mean_squared_error(y_test, pred) ** .5)
```
