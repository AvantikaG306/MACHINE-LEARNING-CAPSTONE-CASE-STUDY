# Person 2 - Preprocessing and Visualisation
Build a reusable numeric regression pipeline and visualise relationships in the housing data.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
df = pd.read_csv('housing.csv')
numeric = df.select_dtypes(include='number')
X = SimpleImputer(strategy='median').fit_transform(numeric.drop(columns='SalePrice'))
X = StandardScaler().fit_transform(X)
sns.boxplot(data=df, x='Overall Qual', y='SalePrice'); plt.title('Price by quality'); plt.show()
sns.heatmap(numeric.corr()[['SalePrice']].sort_values('SalePrice'), cmap='coolwarm'); plt.show()
```
