# Person 2 - Lasso and Polynomial Regression
Compare sparse linear modelling with a second-degree feature expansion.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.linear_model import Lasso, LinearRegression
from sklearn.metrics import r2_score
df = pd.read_csv('housing.csv'); X = df.select_dtypes(include='number').drop(columns='SalePrice'); y = df.SalePrice
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=.2, random_state=42)
models = {'Lasso': make_pipeline(SimpleImputer(strategy='median'), StandardScaler(), Lasso(alpha=100)), 'Polynomial': make_pipeline(SimpleImputer(strategy='median'), StandardScaler(), PolynomialFeatures(2), LinearRegression())}
for name, model in models.items():
    model.fit(Xtr, ytr); print(name, r2_score(yte, model.predict(Xte)))
```
