# Person 2 - Gaussian Naive Bayes
Train a probabilistic classifier for online purchase intent.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.naive_bayes import GaussianNB
from sklearn.pipeline import make_pipeline
from sklearn.metrics import classification_report
df = pd.read_csv('online_shoppers_intention.csv'); X = df.select_dtypes(include='number'); y = df.Revenue.astype(int)
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=.2, stratify=y, random_state=42)
model = make_pipeline(SimpleImputer(strategy='median'), StandardScaler(), GaussianNB()).fit(Xtr, ytr)
print(classification_report(yte, model.predict(Xte)))
```
